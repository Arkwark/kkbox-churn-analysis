```python
# 1. Доля оттока в train
print("Доля оттока:")
print(train['is_churn'].value_counts(normalize=True))

# 2. Пропуски в members
print("\nПропуски в members:")
print(members.isnull().sum())

# 3. Период в transactions (важно понять формат даты)
print("\nТип данных даты транзакции:", transactions['transaction_date'].dtype)
print("Мин дата:", transactions['transaction_date'].min())
print("Макс дата:", transactions['transaction_date'].max())

# 4. Размер логов
# Проверка количества уникальных пользователей в логах
unique_users_in_logs = user_logs['msno'].nunique()
total_users_in_train = train['msno'].nunique()
print(f"Уникальных пользователей в логах: {unique_users_in_logs}")
print(f"Уникальных пользователей в train: {total_users_in_train}")
print(f"Доля пользователей из train, у которых есть логи: {unique_users_in_logs / total_users_in_train:.2%}")
```
